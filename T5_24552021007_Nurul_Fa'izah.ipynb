```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload gambar
uploaded = files.upload()

# Membaca gambar
nama_file = list(uploaded.keys())[0]
img = cv2.imread(nama_file)

# Konversi ke Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Menampilkan gambar grayscale
plt.figure(figsize=(5, 5))
plt.imshow(gray, cmap='gray')
plt.title('Citra Grayscale')
plt.axis('off')
plt.show()
```

```python
# @title Kode untuk Eksperimen Nilai Ambang Batas (Thresholding)
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Membaca citra asli yang telah diunggah sebelumnya dan mengonversinya menjadi Grayscale
# Kita menggunakan variabel nama_file dari sel sebelumnya (misal: 'nailong.jpg')
try:
    img = Image.open(nama_file)
except NameError:
    # Jika variabel nama_file belum terdefinisi, gunakan default file yang ada
    img = Image.open('nailong.jpg')

gray_img = img.convert('L')
gray_array = np.array(gray_img)

# 2. Buat tiga variasi nilai ambang batas (threshold): T1 = 80, T2 = 128, T3 = 200
T1 = 80
T2 = 128
T3 = 200

# Lakukan proses thresholding secara manual menggunakan NumPy array (operasi perbandingan)
# Jika nilai piksel > T, ubah jadi 255 (putih), jika tidak ubah jadi 0 (hitam)
thresh1 = np.where(gray_array > T1, 255, 0).astype(np.uint8)
thresh2 = np.where(gray_array > T2, 255, 0).astype(np.uint8)
thresh3 = np.where(gray_array > T3, 255, 0).astype(np.uint8)

# Tampilkan hasil ketiga citra biner tersebut secara berdampingan menggunakan plt.subplots
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Menampilkan biner T = 80
axes[0].imshow(thresh1, cmap='gray', vmin=0, vmax=255)
axes[0].set_title(f"Biner (T = {T1})")
axes[0].axis('off')

# Menampilkan biner T = 128
axes[1].imshow(thresh2, cmap='gray', vmin=0, vmax=255)
axes[1].set_title(f"Biner (T = {T2})")
axes[1].axis('off')

# Menampilkan biner T = 200
axes[2].imshow(thresh3, cmap='gray', vmin=0, vmax=255)
axes[2].set_title(f"Biner (T = {T3})")
axes[2].axis('off')

plt.tight_layout()
plt.show()
```

```python
# @title 2.1.2 Memisahkan dan Mengisolasi Kanal R, G, dan B
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# 1. Membaca gambar yang telah diunggah sebelumnya sebagai RGB
try:
    img = np.array(Image.open(nama_file).convert('RGB'))
except NameError:
    # Jika variabel nama_file belum terdefinisi, gunakan default file yang ada
    img = np.array(Image.open('nailong.jpg').convert('RGB'))

# 2. Membuat salinan array untuk mengisolasi setiap kanal warna murni
# (Menyetel kanal lain menjadi 0 agar tampil sesuai warna aslinya, bukan grayscale)
img_R = img.copy()
img_R[:, :, 1] = 0  # Nolkan Hijau
img_R[:, :, 2] = 0  # Nolkan Biru

img_G = img.copy()
img_G[:, :, 0] = 0  # Nolkan Merah
img_G[:, :, 2] = 0  # Nolkan Biru

img_B = img.copy()
img_B[:, :, 0] = 0  # Nolkan Merah
img_B[:, :, 1] = 0  # Nolkan Hijau

# 3. Membuat visualisasi 2 baris dan 2 kolom menggunakan plt.subplots
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Citra RGB asli
axes[0, 0].imshow(img)
axes[0, 0].set_title("Citra Asli (RGB)")

# Kanal Merah murni
axes[0, 1].imshow(img_R)
axes[0, 1].set_title("Hanya Kanal Red (R)")

# Kanal Hijau murni
axes[1, 0].imshow(img_G)
axes[1, 0].set_title("Hanya Kanal Green (G)")

# Kanal Biru murni
axes[1, 1].imshow(img_B)
axes[1, 1].set_title("Hanya Kanal Blue (B)")

# Menyembunyikan sumbu koordinat untuk semua subplot
for ax in axes.ravel():
    ax.axis('off')

plt.tight_layout()
plt.show()
```
