Soal 3 (Scikit-Image - Filtering)

```python
## Jawaban Soal No 3
import cv2
import matplotlib.pyplot as plt
from skimage import color, filters

# Membaca gambar menggunakan OpenCV
gambar_bgr = cv2.imread('download.png')

# Ubah ruang warna dari BGR ke RGB
gambar_rgb = cv2.cvtColor(gambar_bgr, cv2.COLOR_BGR2RGB)

# Slicing array NumPy: [y_min:y_max, x_min:x_max]
gambar_crop = gambar_rgb[200:600, 400:800]

# Konversi gambar hasil crop ke grayscale sebelum deteksi tepi
gambar_gray = color.rgb2gray(gambar_crop)

# Terapkan filter deteksi tepi Sobel (filters.sobel) dari Scikit-Image
tepi_sobel = filters.sobel(gambar_gray)

# Tampilkan hasilnya dengan peta warna (cmap) abu-abu ('gray')
plt.figure(figsize=(6, 6))
plt.imshow(tepi_sobel, cmap='gray')
plt.axis('off')  # Mematikan tampilan sumbu koordinat
plt.title('Deteksi Tepi Sobel (Grayscale)')
plt.show()
```
